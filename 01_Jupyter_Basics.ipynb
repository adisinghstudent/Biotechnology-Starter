## 📂 Understanding File Types

Before we code, let's learn the files you'll see in data science:

*   **`.ipynb` (Interactive Python Notebook):** This file! It contains both code and text.
*   **`.md` (Markdown):** A simple text file for documentation (like the `00_Future_of_Biotech.md` you just read).
*   **`.csv` (Comma Separated Values):** A spreadsheet file. It's just rows and columns of data, separated by commas. We use this to store data.
*   **`.py` (Python Script):** A file containing *only* code. Used for building larger software.

## 🐍 Let's Write Python!

Python is the language of data science. It's designed to be readable, like English.

### 1. Printing
The `print()` command shows you output.

```python
print("Hello, World!")
print("I am learning Biotechnology!")
```

### 2. Variables
Variables are like containers. You can store information in them to use later.

```python
my_name = "Ania"  # This is a string (text)
my_age = 25       # This is a number

print("My name is", my_name)
print("Next year I will be", my_age + 1)
```

### 3. Lists
Lists allow you to store multiple items in one variable. In Biotech, we might store a DNA sequence.

```python
dna_bases = ["A", "C", "T", "G"]

print("The DNA bases are:", dna_bases)
print("The first base is:", dna_bases[0])
```
