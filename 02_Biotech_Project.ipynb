```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# This makes graphs look nicer
sns.set_theme(style="whitegrid")
print("Tools loaded!")
```

## 💾 Step 3: Load the Data

Let's read the CSV file into a Pandas DataFrame (a programmable spreadsheet).

```python
# Load the data
df = pd.read_csv('dataset/data.csv')

# Show the first 5 rows
df.head()
```

## 🔍 Step 4: Explore the Data

Let's see what we have. 
*   `diagnosis`: **M** = Malignant, **B** = Benign.
*   `radius_mean`, `texture_mean`, etc.: Measurements of the cells.

```python
# How many rows and columns?
print("Shape of data:", df.shape)

# Get summary statistics (mean, max, min)
df.describe()
```

### How many of each diagnosis?
Let's count how many Malignant vs Benign cases we have.

```python
df['diagnosis'].value_counts()
```

## 📊 Step 5: Visualize!

Data is easier to understand with pictures. Let's compare the **Radius** (size) of the tumors for Benign vs Malignant.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='radius_mean', hue='diagnosis', kde=True)
plt.title('Tumor Radius: Malignant (M) vs Benign (B)')
plt.show()
```

### Scatter Plot
Do larger tumors also have different textures? Let's plot **Radius** vs **Texture**.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='radius_mean', y='texture_mean', hue='diagnosis')
plt.title('Radius vs Texture')
plt.show()
```

## 🏆 Conclusion

Look at the graphs! 
*   **Malignant** tumors (Orange/M) tend to be **larger** (higher radius) than Benign ones.
*   They also seem to have slightly different textures.

This is exactly what AI models look for. By feeding thousands of these examples to a computer, it learns to draw that line between Blue and Orange automatically.
